```python
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height

    @property
    def width(self):
        return self._width

    @width.setter
    def width(self,width):
        if width <=0:
            raise ValueError('Width must be always be positive')
        else:
            self._width = width
    @property
    def height(self):
        return self._height
        
    @height.setter
    def height(self,height):
        if height <=0:
            raise ValueError('Width must be always be positive')
        else:
            self._height = height
    
    def __str__(self):
        return 'Rectangle(Width:{0},Height:{1})@{2}'\
            .format(self.width,self.height,hex(id(self)))

    def __repr__(self):
        return 'Rectangle(Width:{0},Height:{1})@{2}'\
            .format(self.width,self.height,hex(id(self)))

    def __eq__(self,other):
        return isinstance(other,Rectangle) and \
        self.width == other.width and self.height == other.height
        
    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

print(f'2.1. Calling __repr__ on {Rectangle(10,10).__repr__()}')
print(f'2.1. Calling __str__ on {Rectangle(10,10).__str__()}')

import ctypes
import sys

rectangle_first = Rectangle(100,500)
rectangle_second = rectangle_first
print(f'2.2.reference count using ctypes:{ctypes.c_long.from_address(id(rectangle_first)).value}, \
reference count using sys: {sys.getrefcount(rectangle_first)-1}')
```
